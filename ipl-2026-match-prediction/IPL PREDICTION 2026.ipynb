```python
import pandas as pd

df1 = pd.read_excel(r"C:\Users\MUHAMMED ADHIL P\TINOS\IPL\IPL_2020.xlsx")
df2 = pd.read_excel(r"C:\Users\MUHAMMED ADHIL P\TINOS\IPL\IPL_2021.xlsx")
df3 = pd.read_excel(r"C:\Users\MUHAMMED ADHIL P\TINOS\IPL\IPL_2022.xlsx")
df4 = pd.read_excel(r"C:\Users\MUHAMMED ADHIL P\TINOS\IPL\IPL_2023.xlsx")
df5 = pd.read_excel(r"C:\Users\MUHAMMED ADHIL P\TINOS\IPL\IPL_2024.xlsx")
df6 = pd.read_excel(r"C:\Users\MUHAMMED ADHIL P\TINOS\IPL\IPL_202.xlsx")

df = pd.concat([df1, df2, df3, df4, df5, df6])
```

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.ensemble import RandomForestClassifier

le = LabelEncoder()

df['team1'] = le.fit_transform(df['team1'])
df['team2'] = le.transform(df['team2'])
df['winner'] = le.transform(df['winner'])

X = df[['team1','team2']]
y = df['winner']

model = RandomForestClassifier()
model.fit(X, y)
```

```python
teams = list(le.classes_)
print(teams)
```

```python
teams = [
    'Chennai Super Kings',  
    'Delhi Capitals', 
    'Gujarat Titans', 
    'Kolkata Knight Riders', 
    'Lucknow Super Giants', 
    'Mumbai Indians', 
    'Punjab Kings', 
    'Rajasthan Royals', 
    'Royal Challengers Bangalore', 
    'Sunrisers Hyderabad'
]
```

```python
matches = []

for team1 in teams:
    for team2 in teams:
        if team1 != team2:
            matches.append([team1, team2])

new_matches = pd.DataFrame(matches, columns=['team1','team2'])
```

```python
new_matches['team1'] = le.transform(new_matches['team1'])
new_matches['team2'] = le.transform(new_matches['team2'])

pred = model.predict(new_matches)

winners = le.inverse_transform(pred)
```

```python
result = pd.DataFrame({
    'team1': le.inverse_transform(new_matches['team1']),
    'team2': le.inverse_transform(new_matches['team2']),
    'predicted_winner': winners
})

print(result)
```
